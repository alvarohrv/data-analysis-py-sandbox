Un escalar es un único valor numérico que representa “cuánto” de algo, pero no “hacia dónde”.  
En álgebra lineal, ese valor sirve como factor para afectar otros objetos numéricos (vectores y matrices) sin cambiar su estructura, solo sus valores.

````python
# COPIAS vs VISTAS

# Demostración de que las variables pueden ser referencias al mismo objeto en memoria.
aa = np.array([2, 3, 7, 8, 5])
bb = aa  # bb es una referencia al mismo arreglo que aa
print(id(bb) == id(aa))  #True # bb es una referencia al mismo arreglo que aa
bb[2] = 33  # Modificar el tercer elemento a través de bb
print(aa)  # [ 2  3 33  8  5]  # aa refleja el cambio hecho a través de bb  
aa[0] = 0  # Modificar el primer elemento a través de aa
print(bb)  # [ 0  3 33  8  5]  # bb refleja el cambio hecho a través de aa
print(bb is aa)  #True # ambas variables apuntan al mismo objeto en memoria.

print("------------")
# Creando una copia independiente del arreglo
cc = bb.copy()  # cc es una copia independiente de aa
cc[2] = 55  # Modificar el tercer elemento de cc
print(cc is bb)  # False  # cc es una copia independiente de bb
print(bb)  # [ 0  3 33  8  5]  # bb no se ve afectado por el cambio en cc
print(cc)  # [ 0  3 55  8  5]  # cc es independiente de bb
print("------------")


# Por defecto, las operaciones de slicing y las funciones que crean nuevos arreglos devuelven vistas (views) del arreglo original, no copias independientes.
# Una vista es una referencia al mismo bloque de memoria que el arreglo original, por lo que los cambios en la vista afectan al original y viceversa.
# Si deseas crear una copia independiente del arreglo, puedes usar el método .copy().



# b = a[1:4]  # Vista del arreglo original (elementos desde el índice 1 hasta el 3)
# b[0] = 99   # Modificar el primer elemento de la vista
# print(a)  # [ 2 99  7  8  5]
# # Para crear una copia independiente:
# d = a[1:4].copy()  # Copia independiente del arreglo original
# d[0] = 55          # Modificar el primer elemento de la copia
# print(a)  # [ 2 99  7  8  5]                
````

````python
# Creacion y comportamiento de vistas en NumPy

aa = np.array([2, 3, 7, 8, 5])
bb = aa  ## bb es un alias o una referencia al objeto aa. Son exactamente el mismo objeto en memoria
# Formas de crear una vista:
cc = aa.view()  # bb es una vista del arreglo aa
cc = aa[:]  # bb es una vista del arreglo aa
print("------")
print(id(bb) == id(aa)) # True # bb es una referencia al mismo arreglo que aa
print(id(cc) == id(aa)) # False # cc es una vista, no una referencia directa (pero tampoco es una copia independiente)
print("------")
vw1_a = aa.view()  # vw_a es una vista del arreglo aa
vw2_a = aa.view()  # vw_a es una vista(2) del arreglo aa
vw1_b = bb.view()  # vw_a es una vista del arreglo bb (que es una referencia a aa)
# nota pero aa y bb siguen siendo referencias al mismo arreglo
print(vw1_a is aa)  # False  # vw_a es una vista, no una referencia directa
print(vw1_b is bb)  # False  # vw_a es una vista, no una referencia directa
# ...
print("--- Modificando el arreglo original ---")
aa[-1] = 99  # Modificar el último elemento del arreglo original
# Afecta a todas las vistas y referencias al mismo bloque de memoria
print(f"aa: {aa}")  
print(f"bb: {bb}") 
print(f"vw1_a: {vw1_a}")
print(f"vw2_a: {vw2_a}")
print(f"vw1_b: {vw1_b}")

print("--- Modificando una vista ---")
vw1_b[-2] = 55  
# Afecta a todas las vistas y referencias al mismo bloque de memoria
print(f"aa: {aa}")  
print(f"bb: {bb}") 
print(f"vw1_a: {vw1_a}")
print(f"vw2_a: {vw2_a}")
print(f"vw1_b: {vw1_b}")

# La principal ventaja de las vistas es la eficiencia, especialmente en operaciones complejas de reestructuración de datos, sin necesidad de copiar gigabytes de información.
# Las vistas ahorran una cantidad masiva de memoria en comparación con una copia completa (.copy()).

# Copia Completa (.copy()): Crea un bloque de datos nuevo en otra parte de la RAM. (Ocupa el doble de espacio).
# Vista (.view()): Solo crea un pequeño encabezado nuevo con metadatos. El espacio extra es insignificante (unos pocos bytes).
````

````python
# 
'''
Regresión lineal (modelo predictivo)

REGRESIÓN LINEAL — MODELO PREDICTIVO
La regresión lineal permite estimar el valor de una variable
(en este caso el precio) a partir de otras variables.

Entre más variables, se necesitas más observaciones para poder estimarlas correctamente.
Aunque normalmente no se busca una solución exacta, sino un modelo que estime los precios.
Más variables → necesitas más datos
Datos reales → modelo estadístico
Ej:
variables independientes#
estrato (2, 3, 4) # Estrato socioeconómico 4 tiene mas Valor
conectividad (1, 2, 3) # Clasificacion 3 tiene mas valor
número de cuartos # En general, más cuartos → más valor
número de baños # En general, más baños → más valor
área (m²) # En general, más área → más valor

variables independientes:
Precio
'''
# LOGICA:
'''
Variables numéricas (se usan directamente)
✔ estrato → 2, 3, 4
✔ cuartos → 1, 2
✔ baños → 1, 2
✔ área → 45, 60, 80…

Cuándo normalizar, se recomendable si:
hay rangos muy diferentes
usarás machine learning
usarás gradiente descendente

El modelo usará:
Precio = b0 + b1(estrato) + b2(conectividad) + b3(cuartos) + b4(baños) + b5(area) 
En forma matricial:
AX = B
donde:
A = variables
X = coeficientes que queremos encontrar
B = precios reales

## Ejemplo con 10 observaciones (Bogotá) (Se recomienda +50)
### 📊 Datos de ejemplo
| Estrato | Conectividad | Cuartos | Baños | Área | Precio (millones COP) |
| ------- | ------------ | ------- | ----- | ---- | --------------------- |
| 2       | 1            | 1       | 1     | 45   | 120                   |
| 2       | 2            | 2       | 1     | 55   | 150                   |
| 3       | 2            | 2       | 2     | 65   | 210                   |
| 3       | 3            | 2       | 2     | 70   | 240                   |
| 4       | 3            | 3       | 2     | 90   | 360                   |
| 4       | 2            | 3       | 2     | 85   | 330                   |
| 3       | 1            | 1       | 1     | 50   | 140                   |
| 2       | 1            | 2       | 1     | 60   | 160                   |
| 4       | 3            | 3       | 3     | 110  | 480                   |
| 3       | 2            | 2       | 1     | 75   | 220                   |
'''

# Variables independientes
A = np.array([
    [2,1,1,1,45],
    [2,2,2,1,55],
    [3,2,2,2,65],
    [3,3,2,2,70],
    [4,3,3,2,90],
    [4,2,3,2,85],
    [3,1,1,1,50],
    [2,1,2,1,60],
    [4,3,3,3,110],
    [3,2,2,1,75]
])

# Precios reales
B = np.array([120,150,210,240,360,330,140,160,480,220])

# Añadimos columna de 1s para el intercepto (b0)
unos = np.ones((A.shape[0], 1))
A = np.hstack((unos, A))

# Resolver usando mínimos cuadrados
coeficientes, _, _, _ = np.linalg.lstsq(A, B, rcond=None)





print("Coeficientes del modelo:")
print(coeficientes)

# 
'''
Regresión lineal (modelo predictivo)
Entre más variables, se necesitas más observaciones para poder estimarlas correctamente.
Aunque normalmente no se busca una solución exacta, sino un modelo que estime los precios.
Más variables → necesitas más datos
Datos reales → modelo estadístico
Ej:
variables independientes#
estrato (2, 3, 4) # Estrato socioeconómico 4 tiene mas Valor
conectividad (1, 2, 3) # Clasificacion 3 tiene mas valor
número de cuartos # En general, más cuartos → más valor
número de baños # En general, más baños → más valor
área (m²) # En general, más área → más valor

variables independientes:
Precio
'''
# LOGICA:
'''
Variables numéricas (se usan directamente)
✔ estrato → 2, 3, 4
✔ cuartos → 1, 2
✔ baños → 1, 2
✔ área → 45, 60, 80…

Cuándo normalizar, se recomendable si:
hay rangos muy diferentes
usarás machine learning
usarás gradiente descendente

MODELO MATEMÁTICO:

Precio = b0 + b1(estrato) + b2(conectividad) + b3(cuartos) + b4(baños) + b5(area) 
En forma matricial:
AX = B
donde:
A = variables
X = coeficientes que queremos encontrar
B = precios reales

## Ejemplo con 10 observaciones (Bogotá) (Se recomienda +50)
### 📊 Datos de ejemplo
| Estrato | Conectividad | Cuartos | Baños | Área | Precio (millones COP) |
| ------- | ------------ | ------- | ----- | ---- | --------------------- |
| 2       | 1            | 1       | 1     | 45   | 120                   |
| 2       | 2            | 2       | 1     | 55   | 150                   |
| 3       | 2            | 2       | 2     | 65   | 210                   |
| 3       | 3            | 2       | 2     | 70   | 240                   |
| 4       | 3            | 3       | 2     | 90   | 360                   |
| 4       | 2            | 3       | 2     | 85   | 330                   |
| 3       | 1            | 1       | 1     | 50   | 140                   |
| 2       | 1            | 2       | 1     | 60   | 160                   |
| 4       | 3            | 3       | 3     | 110  | 480                   |
| 3       | 2            | 2       | 1     | 75   | 220                   |
'''

# -------------------------
# Variables independientes
# columnas:
# estrato, conectividad, cuartos, baños, área
# -------------------------

A = np.array([
    [2,1,1,1,45],
    [2,2,2,1,55],
    [3,2,2,2,65],
    [3,3,2,2,70],
    [4,3,3,2,90],
    [4,2,3,2,85],
    [3,1,1,1,50],
    [2,1,2,1,60],
    [4,3,3,3,110],
    [3,2,2,1,75]
])

# Precios reales (millones COP)
B = np.array([120,150,210,240,360,330,140,160,480,220])

# Añadimos columna de 1s para el intercepto (b0)
unos = np.ones((A.shape[0], 1))
# print("Columna de unos para el intercepto:")
print(unos)
A = np.hstack((unos, A))
# print("\nMatriz A después de agregar la columna de unos:")
# print(A)

# Resolver usando mínimos cuadrados
coeficientes, _, _, _ = np.linalg.lstsq(A, B, rcond=None)

print("Coeficientes del modelo:")
print(coeficientes)
# Coeficientes del modelo:
# [-148.687   11.478   -6.582   -1.045   38.284    4.388 ]
# [    b0  , estrato ,conectiv, cuartos , baños  ,   m^2 ]
'''
¿Qué obtienes con esto?

b0 = valor base del modelo
b1 = aumento por cada nivel de estrato
b2 = impacto de conectividad
b3 = impacto por cuartos
b4 = impacto por baños
b5 = aumento por cada m²

Interpretación:
• Cada baño adicional aumenta MUCHO el valor
• Estrato aumenta el valor
• Cada m² aumenta el precio
• Algunos coeficientes pueden ser negativos si
  los datos son pocos o inconsistentes
'''

# PREDICCIÓN DE UN NUEVO APARTAMENTO
'''
Ejemplo:
estrato=3
conectividad=2
cuartos=2
baños=1
área=80
'''

print("\nPredicción para un nuevo apartamento:")
print("Características: estrato=3, conectividad=2, cuartos=2, baños=1, área=80 m²")
nuevo = np.array([1, 3, 2, 2, 1, 80])  # incluye el 1 del intercepto
precio_estimado = nuevo @ coeficientes
print("\n---Precio estimado:---")
print(round(precio_estimado, 2), "millones COP")




````
