**Normalización BASICA de texto**

```python
# Total Ventas está como object (texto) y queremos convertirlo a numérico (float) para poder hacer cálculos con él, por tanto:

# copia- nulos antes de intentar la conversión
nulos_originales = df["Total Ventas"].isna() # PARA VALIDACION

# Quitamos el signo $ y las comas (si las hubiera para miles)
# Usamos .replace con regex=True para que busque los símbolos en cualquier parte del texto
df["Total Ventas"] = (
    df["Total Ventas"]
    # .str.replace(r"[\$,€£]", "", regex=True) # Quita $ y ,
    .str.replace(r"[\$,€£]", "", regex=True) # Quita $ y , (y otros símbolos si los hubiera)
        # regex es necesario para que busque el símbolo en cualquier parte del texto, no solo al principio o al final.
        # valor 'True' es necesario para que reemplace todas las ocurrencias, no solo la primera.
        # valor 'False' haría que solo reemplace la primera ocurrencia.
    .str.replace(r"[oOsS]", "0", regex=True)   # CAMBIA 'o' u 'O' por '0'
    .str.strip()                          # Quita espacios laterales
)

# Convertimos a número (float) RED DE SEGURIDAD - Pero validar casos como 'o' en lugar de '0'
df["Total Ventas"] = pd.to_numeric(df["Total Ventas"], errors='coerce')
# errors='coerce' es vital: si encuentra algo que no puede convertir (ej. "null"), 
# en lugar de dar error, lo transforma en NaN.

# Identificamos los "Nuevos Nulos" (los que fallaron por caracteres extraños)
fallas_conversion = df[df["Total Ventas"].isna() & ~nulos_originales]
columnas_interes = ["ID", "Producto", "Ventas", "Precio Unitario", "Total Ventas"]

if not fallas_conversion.empty:
    print("❌ REGISTROS QUE NO SE PUDIERON CONVERTIR:")
    print(fallas_conversion[columnas_interes])
    print(df["Total Ventas"].dtype)
else:
    print("✅ ¡Éxito! Todos los registros se convirtieron correctamente.")
    # 3. Verificamos el resultado
    print(df["Total Ventas"].dtype)
```

**Normalización BASICA de texto**

```python

# 1. Definimos las columnas que queremos limpiar
cols_a_limpiar = ["Producto", "Categoría", "Región"]

for col in cols_a_limpiar:
    # A. Quitamos espacios al inicio y al final
    df[col] = df[col].str.strip() # eliminar espacios al inicio y al final de cada valor en la columna
    # df[col].replace("", np.nan, inplace=True) # No usamos esta línea porque ya la lógica de detección de celdas basura lo hará en el siguiente paso.

    # B. Identificamos celdas que NO tienen contenido real (letras o números)
    # Usamos tu lógica del operador de negación (~) y el accesor .str
    # Esto atrapará celdas con solo espacios, puntos, o vacías
    mascara_basura = ~df[col].str.contains(r"[a-zA-Z0-9]", na=False)
        # ~ → operador de negación para seleccionar filas que NO cumplen la condición
        # na=False → para evitar que los NaN se consideren como True (ya que no contienen letras ni números)
        # df["Producto"].str, le esta diciendo a Pandas: "Trata cada celda de esta columna como si fuera un string".

    # C. Asignamos NaN a esas celdas detectadas con problemas
    df.loc[mascara_basura, col] = np.nan

# 2. Opcional: Convertir todo a minúsculas para normalizar 
for col in cols_a_limpiar:
    df[col] = df[col].str.lower()

print("¡Limpieza completada!")


#Nota: En Python puro, las cadenas de texto (str) tienen sus propios métodos, pero contains no es uno de ellos (en Python estándar usarías el operador in). Sin embargo, en Pandas existe el Accesor .str, que es como un portal que te permite aplicar métodos de strings (y algunos extras de Regex) a toda una columna de una sola vez.
```

**Limpieza de Categoria**

```python
# ---------- 0️⃣ PREPARACIÓN: IDENTIFICAR Y GUARDAR ÍNDICES ----------

# Buscamos las filas donde 'Categoría' es nula
indices_nulos_cat = df[df["Categoría"].isna()].index
print(f"Se encontraron {len(indices_nulos_cat)} filas con Categoría nula.")
print("\n--- FILAS ANTES DE LA LIMPIEZA (Solo las afectadas) ---")
print(df.loc[indices_nulos_cat])


# ---------- 1️⃣ TU CÓDIGO DE LIMPIEZA ----------

# Aplicamos la lógica de completar por producto o poner 'Desconocido'
df["Categoría"] = df.groupby("Producto")["Categoría"]\
                    .transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Desconocido"))

# Si aún quedan vacíos (casos donde no hubo coincidencia)
# df["Categoría"].fillna("Desconocido", inplace=True) # Descontinuado 
df["Categoría"] = df["Categoría"].fillna("Desconocido")
# Red de seguridad para cualquier caso que se nos haya pasado:
# ✔ productos únicos sin categoría
# ✔ datos corruptos
# ✔ inconsistencias
# fillna() es un método de Pandas que se utiliza para rellenar los valores faltantes (NaN) en un DataFrame o Serie. En este caso, se está utilizando para asegurarse de que cualquier valor faltante en la columna 'Categoría' se rellene con "Desconocido" después de intentar completar por producto.

"""
df.groupby("Producto") -> Agrupa filas que tienen el mismo producto.
Grupo: Mouse:
Accesorios
NaN
Accesorios
df.groupby("Producto")["Categoría"] -> Ahora cada grupo solo contiene esa columna.
transform(): ->
✔ aplica una función a cada grupo
✔ devuelve el resultado con el mismo tamaño del DataFrame
✔ permite reasignarlo a la columna original
Sin transform(), perderías la forma original (n°filas, etc).
ambda x: ...  ->  x representa la columna Categoría dentro de cada grupo.
x.fillna(...) -> Rellena los valores faltantes dentro del grupo.
x.mode()[0] -> valor usamos para rellenar
mode() -> Devuelve una serie, tomamos el valor más frecuente.
if not x.mode().empty  -> Rellenar
Si está vacío:  ->   else "Desconocido"
"""

# ---------- 2️⃣ VERIFICACIÓN: COMPARAR EL DESPUÉS ----------

print("\n--- FILAS DESPUÉS DE LA LIMPIEZA (Mismos índices) ---")
# Usamos .loc con los índices que guardamos al principio
print(df.loc[indices_nulos_cat])
# df[df["Categoría"] == "Desconocido"]
```

```python
# ¿Calcular por Región los Ingresos Totales Recalculados a partir de Ventas y Precio Unitario?

# 1. Definimos la función adaptada a tus nombres de columnas
def calcular_ingreso_total(grupo):
    # Multiplicamos la columna de unidades por la de precio y sumamos el resultado
    return (grupo['Ventas'] * grupo['Precio Unitario']).sum()

# 2. Aplicamos el groupby por 'Región'
# En lugar de 'Country', usamos 'Región'
ingreso_por_region = df.groupby('Región').apply(calcular_ingreso_total, include_groups=False)

print("Ingreso Total Recalculado por Región:")
print(ingreso_por_region)

```

```python
# ¿Cual es el top 3 de mejores Regiones en ingresos totales ( Pero indicar en otra columna el total de ventas por región)? 
# 1. Calculamos el ingreso total por región usando la función que definimos antes
ingreso_por_region = df.groupby('Región').apply(calcular_ingreso_total, include_groups=False)   
# 2. Obtenemos el total de ventas por región
ventas_por_region = df.groupby('Región')['Ventas'].sum()
# 3. Combinamos ambos resultados en un DataFrame
resultados = ingreso_por_region.to_frame(name='Ingreso Total').join(ventas_por_region.to_frame(name='Ventas Totales'))
# 4. Obtenemos el top 3 de regiones por ingreso total
top_3_regiones = resultados.sort_values(by='Ingreso Total', ascending=False).head(3)
print("Top 3 de Regiones por Ingreso Total Recalculado:")
print(top_3_regiones)
print("____"*10)

# ¿Cual es el top 3 de peores Regiones en ingresos totales ( Pero indicar en otra columna el total de ventas por región)? 
top_3_peores_regiones = resultados.sort_values(by='Ingreso Total', ascending=True).head(3)
print("Top 3 de Peores Regiones por Ingreso Total Recalculado:")
print(top_3_peores_regiones)
```

#### 1. ¿Por qué existe el `NaN` y qué tipo de dato es?

`NaN` (Not a Number) técnicamente es un número.
* ¿Por qué es un número? En computación, para que una columna de precios sea rápida de procesar, todos los elementos deben ser del mismo tipo (ej. `float64`). Si pusiéramos un string, la columna se volvería lenta.
`type(np.nan)`, es `<class 'float'>`. Es un "marcador de posición" numérico que dice: *"Aquí debería ir un número, pero no lo tenemos"*.

#### 2. El misterio del `None` en las Bases de Datos
¿Puede una base de datos venir con `None`? Sí y no.
* En la Base de Datos (SQL): Se llama `NULL`. Es un estado que indica que el valor es desconocido o no aplicable.
* Al importar a Python: El conector de la base de datos traduce ese `NULL` de SQL al `None` de Python.
* ¿es "grave"? `None` es un objeto de tipo `NoneType`.
No puedes sumar `5 + None` = ❌ (daría error) rompe la operación: `10 + None = Error`
Pero pandas lo convierte a NaN para evitar errores.
Pues sí puedes sumar `5 + NaN` (el resultado es `NaN`). Por eso Pandas prefiere convertir los `None` a `NaN`: para que el código no "explote" al hacer cálculos.
