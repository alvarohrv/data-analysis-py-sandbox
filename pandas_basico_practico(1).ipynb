**Limpieza de Categoria**

```python
# ---------- 0️⃣ PREPARACIÓN: IDENTIFICAR Y GUARDAR ÍNDICES ----------

# Buscamos las filas donde 'Categoría' es nula
indices_nulos_cat = df[df["Categoría"].isna()].index
print(f"Se encontraron {len(indices_nulos_cat)} filas con Categoría nula.")
print("\n--- FILAS ANTES DE LA LIMPIEZA (Solo las afectadas) ---")
print(df.loc[indices_nulos_cat])


# ---------- 1️⃣ TU CÓDIGO DE LIMPIEZA ----------

# Aplicamos la lógica de completar por producto o poner 'Desconocido'
df["Categoría"] = df.groupby("Producto")["Categoría"]\
                    .transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Desconocido"))

# Si aún quedan vacíos (casos donde no hubo coincidencia)
# df["Categoría"].fillna("Desconocido", inplace=True) # Descontinuado 
df["Categoría"] = df["Categoría"].fillna("Desconocido")
# Red de seguridad para cualquier caso que se nos haya pasado:
# ✔ productos únicos sin categoría
# ✔ datos corruptos
# ✔ inconsistencias

"""
df.groupby("Producto") -> Agrupa filas que tienen el mismo producto.
Grupo: Mouse:
Accesorios
NaN
Accesorios
df.groupby("Producto")["Categoría"] -> Ahora cada grupo solo contiene esa columna.
transform(): ->
✔ aplica una función a cada grupo
✔ devuelve el resultado con el mismo tamaño del DataFrame
✔ permite reasignarlo a la columna original
Sin transform(), perderías la forma original (n°filas, etc).
ambda x: ...  ->  x representa la columna Categoría dentro de cada grupo.
x.fillna(...) -> Rellena los valores faltantes dentro del grupo.
x.mode()[0] -> valor usamos para rellenar
mode() -> Devuelve una serie, tomamos el valor más frecuente.
if not x.mode().empty  -> Rellenar
Si está vacío:  ->   else "Desconocido"
"""

# ---------- 2️⃣ VERIFICACIÓN: COMPARAR EL DESPUÉS ----------

print("\n--- FILAS DESPUÉS DE LA LIMPIEZA (Mismos índices) ---")
# Usamos .loc con los índices que guardamos al principio
print(df.loc[indices_nulos_cat])
# df[df["Categoría"] == "Desconocido"]
```

**LIMPIAR PRECIO UNITARIO**

```python
# ---------- 0️⃣ PREPARACIÓN: IDENTIFICAR Y GUARDAR ÍNDICES ----------

# Buscamos las filas donde 'Precio Unitario' es nula
indices_nulos_cat = df[df["Precio Unitario"].isna()].index
print(f"Se encontraron {len(indices_nulos_cat)} filas con Precio Unitario nula.")
print("\n--- FILAS ANTES DE LA LIMPIEZA (Solo las afectadas) ---")
print(df.loc[indices_nulos_cat])


# ---------- 1️⃣ LIMPIAR PRECIO UNITARIO ----------


# 1️⃣ Calcular precio unitario si faltante pero existen Total_Ventas y Ventas
# Precio Unitario = Total Ventas / Ventas
mask = df["Precio Unitario"].isna() & df["Total Ventas"].notna() & df["Ventas"].notna()
# mask es un filtro lógico que identifica filas donde:
# -falta el precio unitario
# -sí existe el total de ventas
# -sí existe la cantidad vendida
# print("Validando la lógica 'mask':")
# print(mask)
# 18    False
# 19    False
# 20     True
# 21    False
print("\n--- FILAS CON PRECIO UNITARIO NULO PERO TOTAL VENTAS Y VENTAS NO NULOS ---")
print(df.loc[mask])
# --- FILAS CON PRECIO UNITARIO NULO PERO TOTAL VENTAS Y VENTAS NO NULOS ---
#     ID   Producto Categoría  Ventas  Precio Unitario  Total Ventas Región
# 20  22  Micrófono     Audio     8.0              NaN        3200.0  Norte
# 27  29  Impresora   Oficina    10.0              NaN        4000.0    Sur
# ignora:
# --  30  Cable USB  Accesorios     NaN              NaN         800.0  Oeste
df.loc[mask, "Precio Unitario"] = df.loc[mask, "Total Ventas"] / df.loc[mask, "Ventas"]

# 2️⃣ Usar promedio del mismo producto
df["Precio Unitario"] = df.groupby("Producto")["Precio Unitario"]\
    .transform(lambda x: x.fillna(x.mean())) # Uso de la Media para rellenar el precio unitario faltante dentro de cada grupo de producto.

# 3️⃣ NO rellenar con promedio global
# df["Precio Unitario"].fillna(df["Precio Unitario"].mean(), inplace=True)
# df["Precio Unitario"] = df["Precio Unitario"].fillna(df["Precio Unitario"].mean())
# dejamos NaN si no se pudo inferir



# ---------- 2️⃣ VERIFICACIÓN: COMPARAR EL DESPUÉS ----------
print("\n--- FILAS DESPUÉS DE LA LIMPIEZA (Mismos índices) ---")
# Usamos .loc con los índices que guardamos al principio
print(df.loc[indices_nulos_cat])
```

```python
# ---------- 0️⃣ PREPARACIÓN: IDENTIFICAR Y GUARDAR ÍNDICES ----------

# Buscamos las filas donde 'Ventas' es nula
indices_nulos_cat = df[df["Ventas"].isna()].index

print(f"Se encontraron {len(indices_nulos_cat)} filas con Ventas nula.")
print("\n--- FILAS ANTES DE LA LIMPIEZA (Solo las afectadas) ---")
print(df.loc[indices_nulos_cat])

# ---------- 1️⃣ LIMPIAR VENTAS ----------

# Primero arreglamos los casos donde Total Ventas es 0
df.loc[df['Total Ventas'] == 0, 'Ventas'] = df['Ventas'].fillna(0)
#filtro de seguridad para paso posterior
# Ubica las filas donde el dinero recaudado es cero, y valida la columna de cantidad vendida,..
# Si en esa fila hay un NaN,  indicar un '0'

# # Luego calculamos: Ventas = Total / Precio (Solo donde Ventas sea NaN, no cero)
df['Ventas'] = df['Ventas'].fillna(df['Total Ventas'] / df['Precio Unitario'])

# # --- 2️⃣ REGLA DE GRUPO: Promedio por producto ---
# df["Ventas"] = df.groupby("Producto")["Ventas"]\
#                  .transform(lambda x: x.fillna(x.mean()))
# # --- 3️⃣ REGLA DE CIERRE: Último recurso ---
# df["Ventas"] = df["Ventas"].fillna(0)
'''
Antes se había propuesto rellenar con el promedio por producto para mantener la coherencia dentro de cada categoría de producto. Pero decidimos no hacerlo porque:
- Podría introducir sesgos si el producto tiene pocos registros o mucha variabilidad.
Se permite dejar el NaN en este caso, ya que no se puede inferir un valor razonable sin más información. Es mejor tener un NaN que un valor incorrecto.
'''

###### NO USADO 
# # rellenar con promedio del producto
# df["Ventas"] = df.groupby("Producto")["Ventas"]\
#                  .transform(lambda x: x.fillna(x.mean()))
# # si aún queda NaN
# # df["Ventas"].fillna(0, inplace=True)
# df ["Ventas"] = df["Ventas"].fillna(0)  


# ---------- 2️⃣ VERIFICACIÓN: COMPARAR EL DESPUÉS ----------

print("\n--- FILAS DESPUÉS DE LA LIMPIEZA (Mismos índices) ---")
# Usamos .loc con los índices que guardamos al principio
print(df.loc[indices_nulos_cat])



```

### 📝 Nota de Estudio:

#### 1. ¿Por qué existe el `NaN` y qué tipo de dato es?

`NaN` (Not a Number) técnicamente es un número.
* ¿Por qué es un número? En computación, para que una columna de precios sea rápida de procesar, todos los elementos deben ser del mismo tipo (ej. `float64`). Si pusiéramos un string, la columna se volvería lenta.
`type(np.nan)`, es `<class 'float'>`. Es un "marcador de posición" numérico que dice: *"Aquí debería ir un número, pero no lo tenemos"*.

#### 2. El misterio del `None` en las Bases de Datos
¿Puede una base de datos venir con `None`? Sí y no.
* En la Base de Datos (SQL): Se llama `NULL`. Es un estado que indica que el valor es desconocido o no aplicable.
* Al importar a Python: El conector de la base de datos traduce ese `NULL` de SQL al `None` de Python.
* ¿es "grave"? `None` es un objeto de tipo `NoneType`.
No puedes sumar `5 + None` = ❌ (daría error) rompe la operación: `10 + None = Error`
Pero pandas lo convierte a NaN para evitar errores.
Pues sí puedes sumar `5 + NaN` (el resultado es `NaN`). Por eso Pandas prefiere convertir los `None` a `NaN`: para que el código no "explote" al hacer cálculos.


#### 3. ¿Por qué aparecen valores nulos en un dataset?
