 **Tema blog: distribución de probabilidad condicional**

### **Frecuencia absoluta**
Número de veces que aparece una categoría en el conjunto de datos.
Ejemplo: “Adelie aparece 152 veces”.

### **Frecuencia relativa**
Proporción que representa una categoría respecto al total.
Se obtiene dividiendo la frecuencia absoluta entre el total de observaciones.
Ejemplo: “Adelie representa el 44% del total”.

### **Frecuencia relativa condicional**
Proporción de una categoría dentro de un grupo específico.
Ejemplo: “Dentro de la isla Biscoe, el 26% de los pingüinos son Adelie”.

### **Tabla de contingencia**
Tabla de conteos cruzados entre dos variables categóricas.
Cada celda contiene una frecuencia absoluta.
Ejemplo: `species × island`.

### **Tabla de contingencia normalizada**
Tabla de contingencia en la que los conteos han sido divididos por un total.
Puede normalizarse por:
* Filas
* Columnas
* Total global
-Tabla de contingencia normalizada por island: cada conteo se divide por el total de casos de la isla; se obtienen frecuencias relativas condicionales

### **Tabla de contingencia normalizada por fila**
Cada celda se divide por el total de su fila.
Cada fila suma 1.


### **PMF (Función de Masa de Probabilidad) (Probabilidad)**
Función que asigna una probabilidad a cada valor posible de una variable aleatoria discreta.
En datos reales, se estima mediante frecuencias relativas.

### **PMF empírica**
PMF calculada a partir de datos observados, no de un modelo teórico.
Se obtiene usando frecuencias relativas.
Ejemplo:
```python
data['species'].value_counts(normalize=True)
```

### **PMF categórica**
PMF aplicada a una variable categórica.
Cada categoría tiene asociada una probabilidad.

### **PMF categórica condicional**
PMF de una variable categórica condicionada a otra variable.
Ejemplo:
> ( P(\text{species} \mid \text{island}) )
Cada fila de la tabla corresponde a una PMF válida (suma 1).
-para cada isla, la fila/columna correspondiente es una función de masa de probabilidad sobre species.

### **Distribución categórica (interpretacion)**
Descripción de cómo se reparten las probabilidades entre las categorías de una variable.
Incluye:
* Proporciones
* Categoría dominante (moda)
* Categorías minoritarias
* Balance vs concentración
* Coherencia con el fenómeno estudiado
📌 No es un número, es una interpretación global.

### **Distribución categórica condicional**
Distribución categórica analizada dentro de cada grupo de otra variable.
Ejemplo:
> “Cómo cambia la distribución de species según la island”.


### **Moda**
Categoría con mayor probabilidad en una PMF categórica.
Equivale al valor más frecuente.

### **Cardinalidad**
Número de categorías únicas de una variable categórica.
Mide diversidad, no concentración.

### **Balance / desbalance**
Grado en que las probabilidades están repartidas de forma uniforme o concentradas en pocas categorías.

### **Coherencia de la distribución**
Evaluación de si la distribución observada tiene sentido según el contexto del problema.
Se usa para validación y limpieza de datos.


### **Independencia entre variables categóricas**
Dos variables son independientes si la distribución de una no cambia al condicionar por la otra.

### **Chi-cuadrado de independencia**
Prueba estadística que evalúa si las diferencias entre distribuciones categóricas observadas pueden atribuirse al azar.
