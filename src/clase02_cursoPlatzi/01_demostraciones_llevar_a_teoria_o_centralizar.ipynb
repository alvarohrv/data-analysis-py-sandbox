### Estadística descriptiva para variables categóricas.

Como las variables categóricas no son números (no puedes promediar "Rojo" y "Azul"), la única forma de entenderlas es midiendo su frecuencia:  
Conteos (Frecuencia Absoluta): Es simplemente decir "¿Cuántos hay?". Ejemplo: "Hay 50 manzanas y 30 peras".  
Proporciones (Frecuencia Relativa): Es decir "¿Qué parte del total representan?" (relacion de correspondencia entre las partes y el todo). Se suele expresar en porcentajes. Ejemplo: "62.5% son manzanas y 37.5% son peras".  
la Distribución es la imagen completa que forman esas piezas. Para un analista de datos, no basta con saber que hay "50 manzanas"; lo que realmente importa es cómo están repartidas todas las frutas en el conjunto.   
La Moda es la única medida de tendencia central que funciona aquí (estadística descriptiva): es simplemente la categoría que más se repite.  
La cardinalidad es el número de categorías únicas que tiene una variable. ¿Para qué sirve en la validación? Si esperas que una variable tenga cardinalidad 2 (ej. "Suscrito: Sí/No") y al explorar ves que la cardinalidad es 5, ¡tienes un problema de limpieza! (Probablemente alguien escribió "si", "SI", "sí", "no", "NO").  

En variables categóricas se calculan principalmente:  
Frecuencia (conteo) (que puede visualizarse con tablas o diagramas de barra)  
Proporción / porcentaje (qué tan representativa es cada categoría)
Distribución (Cuando hablamos de distribución, no hablamos de un número, sino de cómo se reparten los valores dentro de un conjunto de datos.)
* En variables numéricas, la distribución se ve como una curva (como la famosa campana de Gauss).
* En variables categóricas, la distribución se visualiza usualmente con un gráfico de barras o un gráfico de pastel.
Moda (El "Líder" de la categoría).
Cardinalidad (La cardinalidad mide la diversidad de los datos y no su reparto. Es simplemente el conteo de los valores únicos.).
No se utilizan medias, promedios ni desviaciones estándar como medidas principales.

- Frecuencia Absoluta -->   ¿Cuántos hay de cada uno?
- Frecuencia Relativa -->   ¿Qué porcentaje representa cada uno del total?  
- Distribución -->   ¿Cómo se reparten los datos en todas las categorías?
- Moda -->   ¿Cuál es el grupo más común?
- Cardinalidad -->   ¿Cuántas etiquetas distintas existen?
- Valores Nulos -->   ¿A cuántos registros les falta la etiqueta?

~~~
EJEMPLO:
Si en tu dataset hay 152 pingüinos de la especie Adelie, el CONTEO es 152.
los 152 pingüinos Adelie es la FRECUENCIA ABSOLUTA respecto al total de la muestra (344 pingüinos),
Y si hay 152 pingüinos Adelie de un total de 344, la FRECUENCIA RELATIVA es 152 / 344 = Aprox 0.44 (o 44%). Decimales entre 0 y 1 (o visto como porcentaje) (Probabilidad/PMF)
~~~

Nota: En la distribucion se indaga:
¿Están todas en una sola canasta o repartidas en muchas?
¿Hay pocas canastas con muchas manzanas o muchas con pocas?
¿La mayoría de las canastas tiene cantidades similares, equitativas o equilibradas?
¿hay categorías con casi cero datos?
¿Cual es la canasta con mas frutas y cual menos?
Eso cambia completamente la interpretación, aunque el total sea el mismo.
- Ejemplo de Distribucion:
Dos empresas tienen:
salario promedio: $2.000
Pero:
Empresa A:
1900, 2000, 2100, 1950, 2050
Empresa B:
500, 500, 500, 500, 6000
-- Mismo promedio
-- Distribuciones muy distintas

Nota: por tanto:
“Cuando haces limpieza y validación podemos verlo como si estuviéramos explorando una variable categórica”
Esto porque cuando haces limpieza y validación, muchas veces no te interesa aún el promedio, la desviación o fórmulas complejas, sino:
¿Qué valores existen?
¿Cuántos hay de cada uno?
¿Hay valores inesperados?
¿Hay nulos?
¿Hay categorías mal escritas?
Eso es exactamente lo que se hace al explorar una variable categórica.

NOTA: en variables categóricas no se mide dispersión en el sentido estadístico clásico (por si se escucha el tema de dispercion en datos categoricos),
No se utilizan medidas como varianza o desviación estándar, ya que no existe un valor central numérico, promedio distancia fisica, ni distancias entre categorías, y esos conceptos miden qué tan lejos está un valor de un centro numérico.
Se analiza en su lugar: concentración (como la moda) vs diversidad (cardinalidad). Por ejemplo:
¿Los datos están concentrados en pocas categorías o repartidos entre muchas?
Esto puede evaluarse mediante:
- Proporciones / porcentajes: permiten identificar categorías con alta concentración (muy representativas) y categorías con baja representatividad.
- Moda: indica si existe una categoría claramente dominante, lo cual sugiere alta concentración.
- Cardinalidad: mide cuántas categorías distintas existen; una cardinalidad alta indica mayor diversidad de categorías, pero no informa cómo se reparten sus frecuencias.
En variables categóricas no medimos qué tan “lejos” están los datos entre sí (dispersión), sino qué tan concentrados o repartidos están dentro del conjunto, utilizando medidas basadas en frecuencias y proporciones.
En categorías no medimos qué tan "lejos" están los datos (dispersión), sino qué tan concentrados o repartidos está

```python
############ Frecuencia (conteo) en VARIABLES CATEGÓRICAS #######
# “¿Cuántos pingüinos hay de cada especie?”
# Frecuencia (conteo) ¿Cuántos hay de cada categoría?
# Usaremos la variable categórica "species".
# Usaremos el método value_counts() de pandas para contar la frecuencia de cada categoría en la columna "species" del DataFrame df.
# freq = data["species"].value_counts()
freq = data.value_counts("species", sort=True) # sort=True realmente es la configuración por defecto, ordena de mayor a menor frecuencia
# mostrar el resultado
print("El resultado muestra el conteo de pingüinos por especie en el conjunto de datos.")
print("----------------")
print(freq)
print("Total de pingüinos:", freq.sum())
```

```python
############# Proporción 
# relación entre las partes y el todo.
#  ¿Cuál es la proporción de cada categoría respecto al total?
#¿Qué tan representativa es cada categoría respecto al total?
prop = data["species"].value_counts(normalize=True)
# prop es un objeto especial de Pandas llamado Series que permite operaciones vectorizadas.
print("El resultado muestra la proporción (en porcentaje) de pingüinos por especie en el conjunto de datos.")
print("----------------")
display(prop * 100) # En porcentaje:
# Nota: El objeto Series en esta operacion toma cada valor de la columna y lo multiplica individualmente por 100, pero mantiene las etiquetas (nombres) donde están.



```

# El objeto Series es exclusivo de la librería Pandas.
Una Series es básicamente una columna inteligente que tiene:
Valores: Los datos (como los porcentajes que viste).
Índice (Index): Las etiquetas o nombres de cada fila (como "Adelie", "Gentoo").
Nombre: La columna misma puede tener un nombre (como "species").
¿Por qué Pandas inventó la Series si ya existían las listas?
Por tres razones que hacen que tu código de los pingüinos funcione:
Alineación automática: Si sumas dos Series, Pandas no suma la posición 1 con la 1; busca que las etiquetas coincidan (ej. suma "Adelie" con "Adelie").
Operaciones vectorizadas: Es lo que te mencioné antes: prop * 100. En una lista normal de Python, eso fallaría o haría algo raro. En una Series, multiplica todo el bloque de datos al instante.
Métodos estadísticos: Las listas de Python no tienen un botón de .value_counts() o .mean(). Pandas creó la Series para que pudieras pedirle estadísticas con una sola palabra.

> 3. ¿Dónde se ubica este dato? (Posición): Cuartiles y Percentiles.
> 4. ¿Qué forma tiene el gráfico? (Forma): Asimetría y Curtosis.

```python
media = peso.mean()
std = peso.std()
lim_inf = media - std
lim_sup = media + std
print("Intervalo de una desviación estándar alrededor de la media:")
print(f"{lim_inf:.2f} g  -  {lim_sup:.2f} g") # 3399.80 g  -  5003.71 g

# Porcentaje real dentro del intervalo
porcentaje_real = ((peso >= lim_inf) & (peso <= lim_sup)).mean() * 100
print(f"Porcentaje real de pingüinos dentro de una desviación estándar: {porcentaje_real:.2f}%")
# Porcentaje real de pingüinos dentro de una desviación estándar: 67.54%

# 📌 Interpretación
print(f"El {porcentaje_real:.2f}% de los pingüinos tiene un peso dentro de una desviación estándar del promedio.")
# El 67.54% de los pingüinos tiene un peso dentro de una desviación estándar del promedio.

# nota: En distribuciones normales contiene cerca del 68% de los valores, pero en la práctica es mejor calcular el porcentaje real para entender la concentración y detectar valores inusuales.
#--------------------------------
# Separar datos
dentro = peso[(peso >= lim_inf) & (peso <= lim_sup)]
fuera = peso[(peso < lim_inf) | (peso > lim_sup)]

plt.hist(dentro, bins=20, alpha=0.7, label="Dentro de ±1σ")
plt.hist(fuera, bins=20, alpha=0.7, label="Fuera de ±1σ")

plt.axvline(media, linestyle="--", label="Media")
plt.axvline(lim_inf, linestyle=":")
plt.axvline(lim_sup, linestyle=":")

plt.title("Peso de pingüinos: media ± desviación estándar")
plt.xlabel("Peso (g)")
plt.ylabel("Frecuencia")
plt.legend()
plt.show()

#--------------------------------

# Tema: 1 desviacion estandar respecto a la media
# (media ± 1×desviacion estandar)

# Es el intervalo alrededor del centro que contiene los valores “típicos” del conjunto.
# valores dentro → “comportamiento normal / esperado”
# valores fuera → “más inusuales” (pero no necesariamente errores a 1xDsv)

# Regla Empírica (68–95–99.7) si los datos siguen una distribución normal:
# Intervalo	% aproximado
# media ± 1σ	≈ 68%
# media ± 2σ	≈ 95%
# media ± 3σ	≈ 99.7%

# NOTA: Cuando usas la Media ± Desviación Estándar, asumes que los datos son simétricos (distribución normal).
```

### Comparación IQR vs Desviación Estándar
- Ambos miden dispersión, pero de formas distintas:  
IQR = Rango Intercuartílico (Q3 - Q1) → mide la dispersión del 50% central de los datos  
Desviación Estándar (σ) → mide la dispersión de todos los datos respecto a la media  
- Inferencias:  
El IQR está desplazado respecto a la media → asimetría  
El intervalo ±σ es más ancho que el IQR → mayor sensibilidad a extremos  
El centro “robusto” (mediana) no coincide con el centro “aritmético” (media)  

**El rango intercuartílico describe el núcleo central y estable de los datos, mientras que la media y la desviación estándar describen la variabilidad alrededor del promedio. Ambos no son equivalentes y responden a preguntas distintas.**

```python
# 3. Medidas de Posición (Localización)

# Concentración de los datos respecto a la media.
kurtosis = peso.kurtosis()
kurtosis
print(f"La curtosis de masa corporal es: {kurtosis:.2f}")
# 📌 Interpretación:
# Alta → distribución puntiaguda (valores muy concentrados respecto a la media)
# media → distribución normal (curtosis = 0)
# Baja / negativa → más plana distribución (valores más dispersos)
# Tipos:
# Mesocúrtica: Distribución homogénea (curtosis = 0).
# Leptocúrtica: Datos muy concentrados (curtosis > 0).
# Platicúrtica: Datos dispersos (curtosis < 0).
# Ejemplo:
# La curtosis de masa corporal es: -0.72
# Baja / negativa → más plana distribución (valores más dispersos)
# Curtosis < 0 → distribución platicúrtica
# Más aplanada que la campana normal
# Menos concentración extrema en el centro
# Colas menos pesadas
# Conclusión: La masa corporal de los pingüinos muestra una distribución ligeramente más plana que la normal, indicando que los pesos están más dispersos y hay menos concentración extrema en el centro.
#Los pesos de los pingüinos están relativamente dispersos alrededor del centro, sin un pico muy pronunciado ni colas extremas.


# Asimetría estadística (Skewness)
skewness = peso.skew()
print(f"La asimetría (skewness) de masa corporal es: {skewness:.2f}")
# 📌 Interpretación:
# Positiva → cola derecha más larga (valores altos extremos)
# Negativa → cola izquierda más larga (valores bajos extremos)
# Ejemplos:
# La asimetría (skewness) de masa corporal es: 0.47
# asimetría leve
# No es una distribución extremadamente sesgada
# Positiva (> 0) → cola hacia la derecha
# Hay algunos pingüinos más pesados que el resto
# Conclusión: La masa corporal de los pingüinos es bastante equilibrada, pero existen algunos individuos más pesados que empujan ligeramente la distribución hacia la derecha (sesgada hacia la derecha).

# Conclusión general sobre la masa corporal:
#A# La masa corporal de los pingüinos presenta una distribución ligeramente más plana que la normal, indicando que los pesos están más dispersos y hay menos concentración extrema en el centro. Aunque la distribución es bastante equilibrada, existen algunos individuos más pesados que empujan ligeramente la distribución hacia la derecha. En general, los pesos de los pingüinos están relativamente dispersos alrededor del centro, sin un pico muy pronunciado ni colas extremas.
#B# La distribución de la masa corporal presenta una asimetría positiva leve (skewness = 0.47), lo que indica la presencia de algunos valores altos. La curtosis negativa (−0.72) sugiere una distribución más aplanada que la normal, sin una concentración extrema alrededor del centro.      
```

```python
# Visualización de media y mediana en el histograma de peso (BINS = segun Freedman-Diaconis) uso de seaborn+matplotlib

# ¿Cuántas barras (bins) debo poner? calculo mediante funcion:
def freedman_diaconis_binwidth(x: pd.Series) -> float:
    """Find optimal binwidth using Freedman-Diaconis rule."""
    IQR = x.quantile(0.75) - x.quantile(0.25)
    N = x.size
    return 2 * IQR / N ** (1 / 3)
# --------
ancho_de_cada_barra = freedman_diaconis_binwidth(data_clean["body_mass_g"])
num_bins = int((data_clean["body_mass_g"].max() - data_clean["body_mass_g"].min()) / ancho_de_cada_barra)
# nota: Si la función freedman_diaconis_binwidth te devolvió 343.19, significa que cada barrita del histograma abarca un rango de 343 gramos de peso de pingüino.
# Los pingüinos pesan entre 2700g y 6300g aproximadamente (un rango total de unos 3600g).Si cada barra mide 343g de ancho: $3600 / 343 son approx 10$ o $11$ barras.

# Histograma con Seaborn usando el ancho de bin calculado:
sns.histplot(
    data=data_clean,
    x="body_mass_g",
    kde=True,  # Agrega una curva de densidad
    binwidth= ancho_de_cada_barra  # Ancho de los bins
    # binwidth= 180,  # Ancho de los bins
)
plt.title("Distribución del peso de los pingüinos con KDE")
plt.xlabel("Peso (g)")
plt.ylabel("Densidad")  
plt.axvline(media_peso, linestyle="--", label="Media", color="red")
plt.axvline(mediana_peso, linestyle=":", label="Mediana")
plt.legend()
#anotacion de pie de pagina
plt.figtext(0.9, 0.02, f"Histograma con {num_bins} barras de { round(ancho_de_cada_barra,2)} gr (Regla de Freedman-Diaconis)", horizontalalignment='right')
# afregar mas espacio en el footer para una linea de texto adicional
plt.subplots_adjust(bottom=0.2)
plt.show()
```
