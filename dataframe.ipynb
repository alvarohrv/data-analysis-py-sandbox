**SELECCIÓN DE DATOS EN PANDAS MEDIANTE `loc`, `iloc` Y NOMBRE DIRECTO**

iloc: Se basa en la posición (número de índice).  
loc: Se basa en las etiquetas de los índices.

```python
## 1. Selección por POSICIÓN → `iloc`
# Se usa cuando quieres acceder por _índices numéricos_ (0, 1, 2…).
print("---- Fila 0, Columna 2")
print(df.iloc[0, 2])        # Fila 0, Columna 2 #→ 'New York'
print("---- Fila 3 completa")
# Fila 3 completa, presentada de forma tubular
F3= df.iloc[3] # df.iloc[3, :] 
# type(F3)  #→ pandas.core.series.Series
print(F3)
print("---- Columna 'Occupation'")
print(df.iloc[:, 3])    # Toda la columna 3 (Occupation)
print("---- Primeras 5 filas, primeras 3 columnas")
print(df.iloc[0:5, 0:3])  # Primeras 5 filas, primeras 3 columnas
print("---- Primeras 5 filas, columnas 0 y 3")
print(df.iloc[0:5, [0, 3]])
print("---- Filas 0, 5 y 15; Columnas 0, 2 y 3")
print(df.iloc[[0, 5, 15], [0, 2, 3]])  # Filas 0, 2 y 4; Columnas 1 y 3

#NOTA:
# El uso de .iloc es necesario porque elimina toda ambigüedad, obligando a pandas a interpretar los argumentos como posiciones numéricas (enteros) y no como nombres o etiquetas (nota, tambien es posible es uso del metodo .loc el cual se usa para acceder a datos por etiquetas o booleanos).
# Por tanto, la forma df[f][c] puede ser ambigua si las filas o columnas tienen etiquetas numéricas que coinciden con los índices posicionales. (en otras palabras, si una fila o columna se llama '0', ¿estás refiriéndote a la fila/columna con etiqueta '0' o a la primera fila/columna en posición 0? .iloc elimina esta ambigüedad al especificar que siempre se trata de posiciones numéricas.)
```

```python
# 2. Selección por ETIQUETA → `loc`
# Se usa cuando quieres acceder por _nombres de filas o columnas_ (etiquetas).
# El primer argumento de .loc se refiere a las filas (etiquetas de índice) y el segundo a las columnas (nombres de columnas).
# loc[] no solo acepta números, también recibe rangos, etiquetas, listas o una lista de condiciones booleanas para seleccionar filas.

# Uso de 'rangos' para seleccionar filas por etiqueta
name_city_occupation_column = df.loc[:,['Name','City', 'Occupation']]
print(" --- DataFrame con columnas 'Name', 'City' y 'Occupation':")
display(name_city_occupation_column)

# uso de 'condiciones booleanas' para seleccionar filas por etiqueta
print(" --- Filas donde la ciudad es 'Chicago' y no sea 'Intern':")
mascara_chicago = df['City'] == 'Chicago' # Esto devuelve una Serie de booleanos (True/False) indicando qué filas cumplen
display(df.loc[mascara_chicago, ['Name', 'City', 'Occupation']])

# Seleccionar filas por etiqueta usando una lista de etiquetas
print(" --- Filas con etiquetas 0, 2 y 5:")
display(df.loc[[0, 2, 5]])

print(" --- Filas donde la edad es mayor a 35:")
display(df.loc[df["Age"] > 35, :])  # filas que cumplen condición
```

```python
## 2. Selección por ETIQUERA  → `loc`
# Se usa cuando quieres acceder por _etiquetas de fila y/o columna.
print("---- Fila 0, Columna 'City': ")
print(df.loc[0, "City"])        # Fila 0, Columna 'City' #→ 'New York'

# Fila 4 completa, presentada de forma tubular?
print("---- Fila 4 completa")
F3 = df.loc[3]  # df.loc[3, :]  # Recordar que la primera fila tiene indice 0
# Nota en algunos diccionarios los indices pueden ser cadenas de texto
# type(F3)  #→ pandas.core.series.Series
print(F3)
print("---- Columna 'Occupation'")
print(df.loc[:, "Occupation"])    # Toda la columna 3 (Occupation)
print("---- Primeras 5 filas, primeras 3 columnas")
print(df.loc[0:4, ["Name", "Age", "City"]])
print("---- Primeras 5 filas, columnas 0 y 3")
print(df.loc[0:4, ["Name", "Occupation"]])
print("---- Filas 0, 5 y 15; Columnas 0, 2 y 3")
print(df.loc[[0, 5, 15], ["Name", "City", "Occupation"]])  # Filas 0, 2 y 4; Columnas 1 y 3
```

```python
# df[...] está diseñado para aceptar distintos tipos de argumentos, y pandas decide qué hacer según el tipo de ese argumento.
# df["Name"] → una sola etiqueta de columna → devuelve una Series con esa columna. (selección simple)
# df[["Name", "Occupation"]] → lista de etiquetas de columna → devuelve un DataFrame con esas columnas.
# df[mask_booleana] → una máscara booleana → devuelve un DataFrame filtrado según la máscara.
### df[mask][['col1', 'col2']] // Indexación directa
```

```python
# 3. Selección simple por NOMBRE DIRECTO (solo columna) parte1
# Se usa cuando quieres acceder por _nombres de columna_ directamente.
### Para obtener **Series** (una sola columna) uso de corchetes simples `[]`
### NOTA: Para obtener **DataFrame** (varias columnas) uso de corchetes dobles

##### output: SERIES
print("---- Columna 'City' directamente")
print(df["City"])    # Toda la columna 'City'
print(type(df["City"]))  #→ pandas.core.series.Series
##### output: DATAFRAME
print("---- Columna 'City' directamente como DataFrame (con cabecera)")
print(df[["City"]])    # Toda la columna 'City'
print(type(df[["City"]]))  #→ pandas.core.frame.DataFrame
```

```python
# La selección simple por NOMBRE DIRECTO permite filtrar filas basadas en condiciones específicas aplicadas a las columnas del DataFrame.
# Esto se logra creando máscaras booleanas que representan las condiciones deseadas y luego aplicándolas al DataFrame para obtener un subconjunto filtrado.
### Ejemplo 1
print(df[df['Age'] >= 39])  # Rows where Age >= 39
print("-----------------------------------------")

### Ejemplo 2 - Múltiples condiciones:
# 1. condición 2: Filtrar por edad mayor a 28
condicion_edad = df['Age'] >= 28
# 2. condición 1: Filtrar por ciudades 'Chicago' o 'Miami'
condicion_ciudades = (df['City'] == 'Chicago') | (df['City'] == 'Miami')
# 3. condición 3: Filtrar por ocupación 'Intern' o 'Developer 
condicion_ocupacion = (df['Occupation'] == 'Intern') | (df['Occupation'] == 'Developer')
# 4. Combina las tres condiciones con AND (&)
# (Ciudades Y Edad Y Ocupación)
resultado = df[condicion_ciudades & condicion_edad & condicion_ocupacion][['Name', 'City', 'Occupation']]
print(resultado)
type(resultado)  #→ pandas.core.frame.DataFrame
```

```python
# 3. Selección por lista de NOMBRES DIRECTOS (solo columnas) parte2
# Se usa cuando quieres acceder por _nombres de columna_ directamente.
### Para obtener **DataFrame** (varias columnas) uso de corchetes dobles
### NOTA: Para obtener **Series** (una sola columna) uso de corchetes simples `[]`
##### output: DataFrame
#print(df[['Name']])  # Select specific columns (retorna un DataFrame y no una Serie)
print("----DataFrame de 2 columnas----")
print(df[["Name", "Occupation"]])  # Columnas 'Name' y 'Occupation'
# equivalente a: df.loc[:, ["Name", "Occupation"]]
# usa el operador de indexación del DataFrame y solo permite seleccionar columnas por nombre cuando se pasa una lista.
#////
# NOTA: para seleccionar a la vez filas y columnas por sus “dos ejes” en la forma df[filas, columnas] se requiere usar uno de los indexadores, iloc o loc, o bien una variante equivalente como df.where o boolean indexing, pero ya no sirve la sintaxis simple df[...] de una sola dimensión.
```

```python
# 🟣 4. Selección avanzada
### ✔ Indexación directa - df[mask][['col1', 'col2']]
### De la forma:
# mascara_booleana = (condicion1) & (condicion2)
# df_filtrado = df[mascara_booleana][["columna"]]
# ----
# print(df[df['Age'] > 28])  # Rows where Age > 28
# print(df[(df['Age'] > 28) & (df['City'] == 'Chicago')])  # Multiple conditions
df[(df['Age'] > 28) & ((df['City'] == 'Chicago') | (df['City'] == 'Miami'))][['Name', 'City', 'Occupation']]
# nota: el operador & representa "AND" y el operador | representa "OR" en pandas.
# nota: la expresion [['Name', 'City', 'Occupation']] se utiliza para seleccionar múltiples columnas específicas del DataFrame resultante después de aplicar las condiciones de filtrado.
# nota: si no se incluye esta parte, el resultado incluirá todos lo campos(columnas) del DataFrame original para las filas que cumplen las condiciones de filtrado.
```
